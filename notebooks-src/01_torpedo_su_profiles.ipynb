# Idealized undrained shear strength profiles

**Educational example; not a capacity model.** We compare the proposed linear relation $s_u(z)=kz$ for three illustrative gradients. Units: $k$ [kPa/m], $z$ [m], $s_u$ [kPa]. The linear law predicts zero strength at the surface; it is a benchmark assumption, not a universal field characterization.

No PLAXIS results or field measurements are used.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

max_depth_m = 30.0
gradients_kPa_per_m = {"Soil A": 1.5, "Soil B": 3.0, "Soil C": 6.0}
z_m = np.linspace(0.0, max_depth_m, 151)
profiles = {name: k * z_m for name, k in gradients_kPa_per_m.items()}
```

```python
summary = pd.DataFrame({"Gradient [kPa/m]": gradients_kPa_per_m, "su at 20 m [kPa]": {name: float(np.interp(20, z_m, su)) for name, su in profiles.items()}})
print(summary.to_string())
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
for name, su in profiles.items():
    ax.plot(su, z_m, label=f"{name} (k={gradients_kPa_per_m[name]:g} kPa/m)")
ax.set_xlabel("Undrained strength, $s_u$ [kPa]")
ax.set_ylabel("Depth, $z$ [m]")
ax.invert_yaxis()
ax.grid(alpha=0.25)
ax.legend()
plt.tight_layout()
plt.show()
```

## Interpretation

Parameter curves can be used to check software input and unit conventions. They **do not determine axial pullout capacity**. The published project should later include source-backed benchmark properties and independent force–displacement comparisons.
