# Estimate undrained shear strength from synthetic CPT data

**Educational synthetic data** only. Conventional approximate equation $s_u=(q_t-\sigma_{v0})/N_{kt}$. Cone correction, soil classification and independent calibration are not covered. In this example $q_t$ is already treated as *corrected* cone resistance. $N_{kt}$ is not a universal constant.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

z_m = np.arange(1., 21., 1.)
qt_kPa = 650. + 80. * z_m + 15. * np.sin(z_m)  # SYNTHETIC, not measured
gamma_total_kN_m3 = 18.0  # synthetic homogeneous assumption
sigma_v0_kPa = gamma_total_kN_m3 * z_m
net_q_kPa = qt_kPa - sigma_v0_kPa
assert np.all(net_q_kPa > 0)
Nkt_cases = [12, 15, 18] # purely illustrative sensitivities
```

```python
table = pd.DataFrame({"depth_m":z_m,"qt_kPa":qt_kPa,"sigma_v0_kPa":sigma_v0_kPa})
for nkt in Nkt_cases:
    table[f"su_Nkt{nkt}_kPa"] = net_q_kPa/nkt
print(table.head(7).to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(7,5))
for nkt in Nkt_cases:
    ax.plot(table[f"su_Nkt{nkt}_kPa"], z_m, label=f"Assumed $N_{{kt}}$={nkt}")
ax.invert_yaxis(); ax.set_xlabel("Estimated undrained strength [kPa]"); ax.set_ylabel("Depth [m]")
ax.grid(alpha=.25); ax.legend()
plt.tight_layout(); plt.show()
```

## Interpretation

Site-specific $N_{kt}$ calibration, appropriate $q_t$ pore-pressure correction and lithological interpretation are necessary before deriving engineering design parameters. No site measurement was processed here.
